## Using Jupytext for Notebook and Script Conversion

This Jupyter Notebook demonstrates how to use Jupytext to convert notebooks to Python scripts and vice versa. It includes steps to install Jupytext, convert a notebook to a Python script, and convert the script back to a notebook. This workflow is useful for version control and collaborative development.

```python
%pip install jupytext
```

```python
# Convert the notebook to a Python script
!jupytext --to py main.ipynb
```

```python
# Convert the Python script back to a notebook
!jupytext --to notebook main.py
```
